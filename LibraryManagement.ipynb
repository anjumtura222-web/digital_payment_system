```python
from abc import ABC,abstractmethod

class LibraryItem(ABC):
    def __init__(self,title):
      self.title=title

@abstractmethod
def display_info(self):
  pass


class Book(LibraryItem):
  def __init__(self,title,author,isbn):
    super().__init__(title)
    self.author=author
    self.isbn=isbn
    self.__is_available= True

  def get_availability(self):
    return self.__is_available
  def set_availability(self,status):
    self.__is_available=status

  def set_availability(self, status):
    self.__is_available = status

  def display_info(self):
    status="Available" if self.get_availability() else "Borrowed"

    print(f"Title:{self.title}")
    print(f"Author:{self.author}")
    print(f"ISBN:{self.isbn}")
    print(f"status:{status}")


class Magazine(LibraryItem):
    def __init__(self, title, issue):
        super().__init__(title)
        self.issue = issue

    def display_info(self):
        print(f"Title:{self.title}")
        print(f"Issue:{self.issue}")


class Member:
   def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.borrowed_books = []

   def borrow_book(self,book):
    if book.get_availability():
      book.set_availability(False)
      self.borrowed_books.append(book)
      print(f"{book.title}successfully borrowed ")
    else:
      print(f"Sorry!{book.title}is already borrowed")

   def return_book(self,book):
    if book in self.borrowed_books:
      book.set_availability(True)
      self.borrowed_books.remove(book)
      print(f"{book.title}sucessfully returned")
    else:
      print(f"You did not borrowed this book")


class Library:
  def __init__(self):
    self.items=[]
    self.members=[]
  def add_item(self,item):
    self.items.append(item)
    print(f"{item.title}added to library")

  def add_member(self, member):
        self.members.append(member)
        print(f"{member.name}registered successfully")

  def show_items(self):
        print("\n===Library Items===")
        for item in self.items:
            item.display_info()
            print("---------")


# main program
library = Library()

book1 = Book(
    "Chokher Bali ",
    "Rabindranath Tagor ",
    "ISB001"
)
book2 = Book(
    "Ghore Baire ",
    "Rabindranath Tagor",
    "ISB001"
)

megazine1 = Magazine(
    "Binibona ",
    "January 2026"
)

library.add_item(book1)
library.add_item(book2)
library.add_item(megazine1)

member1 = Member(
    "Tabassum ",
    "M001"
)

library.add_member(member1)
library.show_items()

print("\n===Borrowing Books===")
member1.borrow_book(book1)
library.show_items()

print("\n===Returning Books===")
member1.return_book(book1)
library.show_items()
```
