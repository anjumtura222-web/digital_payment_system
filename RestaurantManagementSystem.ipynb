```python
from abc import ABC,abstractmethod
class MenuItem(ABC):
  @abstractmethod
  def show_info(self):
    pass

class Food(MenuItem):
  def __init__(self,name,price):
    self.name=name
    self.__price=price

  def show_info(self):
    print("Food:",self.name)
    print("Price:",self.__price)

class Drink(MenuItem):
  def __init__(self,name,price):
    self.name=name
    self.__price=price

  def show_info(self):
    print("Drink:",self.name)
    print("Price:",self.__price)

class Customer:
  def __init__(self,name,code_number):
    self.name=name
    self.code_number=code_number
    self.__orders=[]

  def order_food(self,food):
    self.__orders.append(food)
    print(food.name,"ordered successfully.")

  def cancel_food(self,food):
    if food in self.__orders:
      self.__orders.remove(food)
      print(food.name,"cancelled")
    else:
      print("The food is not in your list.")


  def show_orders(self):
    if len(self.__orders)==0:
      print("No orders yet.")
    else:
      print("\n Your orders:")
      for item in self.__orders:
        item.show_info()
class Restraurent:
  def __init__(self,name):
    self.name=name
    self.menu=[]

  def add_food(self,food):
    self.menu.append(food)

  def show_menu(self):
    print("\n====Restraurent Menu====")
    for item in self.menu:
      item.show_info()


def main():
  restraurent= Restraurent("Food Palace")

  burger=Food("Burger",150)
  pizza=Food("Pizza",400)

  coke=Drink("Mojo",30)
  coffee=Drink("Coffee",100)

  restraurent.add_food(burger)
  restraurent.add_food(pizza)
  restraurent.add_food(coke)
  restraurent.add_food(coffee)

  restraurent.show_menu()

  customer=Customer("John","001")
  customer.order_food(burger)
  customer.order_food(coffee)
  customer.order_food(coke)

  customer.cancel_food(coke)

  customer.show_orders()
if __name__=="__main__":
    main()













```
