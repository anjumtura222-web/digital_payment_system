```python
from abc import ABC,abstractmethod
#abstractclass
class PaymentSystem(ABC):
  @abstractmethod
  def process_payment(self,amount):
   pass

#encapsulation
class Account:

  def __init__(self,account_holder,balance):
    self.account_holder= account_holder
    self.__balance= balance
#getter
  def get_balance(self):
    return self.__balance
#setter
  def set_balance(self,balance):
    self.__balance=balance
#inheritance
class BkashPayment(PaymentSystem):
    def __init__(self,account):
      self.account=account
#methodOverridding
    def process_payment(self, amount):
      service_charge= amount*0.015
      total_amount=amount+service_charge
      if total_amount<=self.account.get_balance():
       new_balance=self.account.get_balance()-total_amount
       self.account.set_balance(new_balance)
       print(f'BkashPayment:{amount:.0f}')
       print(f'service_charge:{service_charge:.0f}')
       print(f'total_amount:{total_amount:.0f}')
       print(f'remaining_balance:{self.account.get_balance():.0f}')
      else:
        print("Insufficient Balance")
#inheritance
class CreditCardPayment(PaymentSystem):
    def __init__(self, account):
        self.account = account
    def process_payment(self, amount):
      if amount>1000:
        discount=amount*0.10
        final_amount=amount-discount
      else:
        discount=0
        final_amount=amount
      if final_amount<=self.account.get_balance():
        new_balance=self.account.get_balance()-final_amount
        self.account.set_balance(new_balance)
        print(f'CreditCardPayment:{amount:.0f}')
        print(f'discount:{discount:.0f}')
        print(f'total:{final_amount:.0f}')
        print(f'remaining_balance:{self.account.get_balance():.0f}')
      else:
        print("Insufficient Balance")
#polymorphism
def make_payment(payment_method,amount):
     payment_method.process_payment(amount)
#testingFlow
user="rabbi"
account = Account(user,5000)
bkash = BkashPayment(account)
credit_card = CreditCardPayment(account)


make_payment(bkash,2000)
print()

make_payment(credit_card,1500)
print()










```
